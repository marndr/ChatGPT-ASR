```python
#import spacy
#from jiwer import compute_measures
#import pandas as pd
#import json
#import os
#import matplotlib.pyplot as plt

SPACY_MODEL = "en_core_web_sm"
nlp = spacy.load(SPACY_MODEL)

def identify_edit_type(ref, asr, llm):
    edit_types = []
    for r, a, l in zip(ref, asr, llm):
        if a == l == r:
            edit_types.append("C")  # left it correct
        elif a != l and l == r:
            edit_types.append("A")  # improve it
        elif a != r and l != r:
            edit_types.append("D")  # left it incorrect
        elif a == r and l != r:
            edit_types.append("B")  # introducing an error
    return edit_types

```

```python
def generate_markdown_table(ref, asr, llm, edit_types):
    md_table = "|  |  |  |  |  |  |  |\n"
    md_table += "|--|--|--|--|--|--|--|\n"
    md_table += "| **REF** | " + " | ".join(ref) + " |\n"
    md_table += "| **ASR** | " + " | ".join(asr) + " |\n"
    md_table += "| **LLM** | " + " | ".join(llm) + " |\n"
    md_table += "| **Type** | " + " | ".join(edit_types) + " |\n"
    return md_table

# Example sentences
reference_transcriptions = ["I", "eat", "***", "salami", "pizza", "salami"]
transcriptions = ["I", "meat", "a", "salami", "pizza", "salami"]
corrected_transcriptions = ["I", "eat", "a", "big", "pizza", "***"]

# Compute edit types
edit_types = identify_edit_type(reference_transcriptions, transcriptions, corrected_transcriptions)

# Generate markdown table
md_table = generate_markdown_table(reference_transcriptions, transcriptions, corrected_transcriptions, edit_types)

# Display markdown table
print(md_table)
```
